```python
source = """
Project name: RockyBot
Purpose: AI news research assistant.
Features: Accepts URLs and documents, retrieves relevant passages,
and answers questions using the supplied information.
Technologies: Python, Streamlit, FAISS, and embeddings.
"""

print(source)
```

```python
supported_answer = """
RockyBot is an AI news research assistant.
It uses Python, Streamlit, FAISS, and embeddings.
It retrieves relevant passages and answers questions
using supplied information.
"""

unsupported_answer = """
RockyBot is an AI news research assistant.
It uses Python, Streamlit, FAISS, and embeddings.
It has 50,000 users and guarantees 100% factual accuracy.
"""
```

```python
def check_claims(source, claims):
    source_lower = source.lower()
    results = []

    for claim in claims:
        if claim.lower() in source_lower:
            status = "SUPPORTED"
        else:
            status = "NEEDS VERIFICATION"

        results.append({
            "claim": claim,
            "status": status
        })

    return results
```

```python
claims = [
    "RockyBot is an AI news research assistant",
    "It uses Python, Streamlit, FAISS, and embeddings",
    "It has 50,000 users",
    "It guarantees 100% factual accuracy"
]

results = check_claims(source, claims)

for item in results:
    print(item["status"], "-", item["claim"])
```

```python
def review_answer(source, answer, claims):
    print("ANSWER:")
    print(answer.strip())

    print("\nCLAIM REVIEW:")

    findings = check_claims(source, claims)

    for finding in findings:
        print(
            f"- {finding['status']}: "
            f"{finding['claim']}"
        )

    return findings
```

```python
claims_to_review = [
    "RockyBot is an AI news research assistant",
    "It has 50,000 users",
    "It guarantees 100% factual accuracy"
]

review_answer(
    source,
    unsupported_answer,
    claims_to_review
)
```
